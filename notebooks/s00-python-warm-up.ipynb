Now run the helper cell below: it powers the ✅ self-checks.

````python
#@title 🧰 Course helpers — run this cell (no need to read it) { display-mode: "form" }
CHECK_RESULTS = {}  # label -> True (correct) / False (not yet) / None (not attempted)

def check(label, test, hint=""):
    """Give instant feedback on an exercise without ever stopping 'Run all'.

    `test` is a function (usually a lambda) that returns True when the answer is right.
    """
    try:
        ok = bool(test())
    except NameError as err:
        CHECK_RESULTS[label] = None
        print(f"⏳ {label}: not attempted yet ({err}).")
        return
    except Exception as err:  # wrong type, wrong shape, missing column, ...
        CHECK_RESULTS[label] = False
        print(f"❌ {label}: checking your answer raised {type(err).__name__}: {err}")
        if hint:
            print(f"   💡 Hint: {hint}")
        return
    CHECK_RESULTS[label] = ok
    if ok:
        print(f"✅ {label}: correct!")
    else:
        print(f"❌ {label}: not yet.")
        if hint:
            print(f"   💡 Hint: {hint}")

def progress(*, return_count=False):
    """Summarise all self-checks run so far; optional 🚀 fast-lane checks are counted separately."""
    core = [v for label, v in CHECK_RESULTS.items() if not label.startswith("🚀")]
    fast = [v for label, v in CHECK_RESULTS.items() if label.startswith("🚀")]
    solved = sum(1 for v in core if v)
    print(f"Solved {solved} of {len(core)} exercises checked so far.")
    if fast:
        print(f"🚀 Fast lane (optional): solved {sum(1 for v in fast if v)} of {len(fast)}.")
    for label, v in CHECK_RESULTS.items():
        print(f"  {'✅' if v else ('⏳' if v is None else '❌')} {label}")
    if return_count:
        return solved

print("Helpers loaded ✔")
````

## 7 · Reading error messages

Read a traceback **from the bottom up**: the last line names the error type and the message, and the arrow `---->` points at the failing line. The six errors you will meet most often:

````python
examples = {
    "NameError":      lambda: undefined_variable + 1,        # a typo, or a cell that was never run
    "TypeError":      lambda: "3" + 4,                       # mixing incompatible types
    "KeyError":       lambda: {"guest": "Ana"}["Guest"],     # a dict key (or DataFrame column) that does not exist
    "IndexError":     lambda: [1, 2, 3][3],                  # a position that does not exist
    "ValueError":     lambda: int("three"),                  # right type, impossible value
    "AttributeError": lambda: [1, 2].upper(),                # a method this type does not have
}
for name, func in examples.items():
    try:
        func()
    except Exception as err:
        print(f"{type(err).__name__:15s} → {err}")
````

**🚀 F2 — Parse messy numbers safely.** Slovenian sources write money as `1.234,50 €`: a dot for thousands, a comma for decimals. A good parser accepts every valid variant and fails *loudly* on garbage, instead of quietly returning a wrong number.

````python
# 🚀 Fast-lane check F2: Write `parse_eur(text)` that turns "1.234,50 €", "12 €", " -3,5 " and "1.000.000" into the floats
#    1234.5, 12.0, -3.5 and 1000000.0, and raises a ValueError with a helpful message for anything else,
#    e.g. "abc", "" or "12,3,4".
# def parse_eur(text):
#     ...
````

````python
#@title ✅ Check — 🚀 F2 { display-mode: "form" }
def _rejects(text):
    try:
        parse_eur(text)
    except ValueError:
        return True
    return False

check("🚀 F2", lambda: parse_eur("1.234,50 €") == 1234.5 and parse_eur("12 €") == 12.0
      and parse_eur(" -3,5 ") == -3.5 and parse_eur("1.000.000") == 1e6
      and all(_rejects(t) for t in ["abc", "", "12,3,4"]),
      hint="Remove the € sign and spaces, check the format (a regular expression helps), then swap the separators before float().")
````

**🚀 F3 — Think in arrays.** pandas is built on **NumPy**, which is fast because it works on whole arrays at once instead of looping in Python. The loop below computes a 7-day moving average of a year of simulated daily temperatures.

````python
import numpy as np

rng = np.random.default_rng(42)
temps = 12 + 10 * np.sin(np.arange(365) / 365 * 2 * np.pi - 1.9) + rng.normal(0, 2, 365)

ma7_loop = [sum(temps[i:i + 7]) / 7 for i in range(len(temps) - 6)]     # one 7-day window at a time
print(len(ma7_loop), "windows; the first one:", round(ma7_loop[0], 2), "°C")
````

````python
# 🚀 Fast-lane check F3: Compute the same moving average WITHOUT a Python loop or comprehension, as a NumPy array `ma7`
#    (length 359). Hint: np.convolve(..., mode="valid") or cumulative sums (np.cumsum).
#    Bonus: compare the speed of both versions with %timeit.
# ma7 = ...
````

````python
#@title ✅ Check — 🚀 F3 { display-mode: "form" }
check('🚀 F3', lambda: isinstance(ma7, np.ndarray) and len(ma7) == 359 and np.allclose(ma7, ma7_loop), hint="np.convolve(temps, np.ones(7) / 7, mode='valid') averages every window of 7 values.")
````
